```python

# Practical 10: Value Iteration in a Simple Grid World

# Step 1: Define transitions
transitions = {
    'A': {'Right': 'B', 'Down': 'C'},
    'B': {'Left': 'A', 'Down': 'G'},
    'C': {'Up': 'A', 'Right': 'G'},
    'G': {}
}

# Step 2: Define rewards
rewards = {
    'A': -1,
    'B': -1,
    'C': -1,
    'G': 10
}

gamma = 0.9

# Step 3: Initialize values
values = {
    'A': 0,
    'B': 0,
    'C': 0,
    'G': 0
}

print("Initial values:")
print(values)

# Step 4: Perform Value Iteration
for i in range(5):
    new_values = values.copy()

    for state in ['A', 'B', 'C']:
        possible_values = []

        for action, next_state in transitions[state].items():
            value = rewards[next_state] + gamma * values[next_state]
            possible_values.append(value)

        new_values[state] = max(possible_values)

    values = new_values

# Step 5: Display final values
print("\nFinal values:")
print(values)

# Step 6: Find the best policy
policy = {}

for state in ['A', 'B', 'C']:
    best_action = None
    best_value = -100

    for action, next_state in transitions[state].items():
        value = rewards[next_state] + gamma * values[next_state]

        if value > best_value:
            best_value = value
            best_action = action

    policy[state] = best_action

# Step 7: Display best policy
print("\nBest policy:")
print(policy)

# Step 8: Display recommended actions
print("\nRecommended Actions:")

for state, action in policy.items():
    print("From", state, "choose", action)
```
