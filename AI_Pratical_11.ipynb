```python

# Practical 11: Propositional Logic and Logical Inference

# Step 1: Store initial facts
facts = {"raining"}

print("Initial facts:", facts)

# Step 2: Store logical rules
rules = {
    "raining": "road_wet",
    "road_wet": "drive_carefully",
    "drive_carefully": "reduce_speed"
}

# Step 3: Perform logical inference using forward chaining
changed = True

while changed:
    changed = False

    for condition, conclusion in rules.items():
        if condition in facts and conclusion not in facts:
            facts.add(conclusion)
            changed = True

# Step 4: Display final conclusions
print("Facts after inference:")

for fact in facts:
    print(fact)
```
